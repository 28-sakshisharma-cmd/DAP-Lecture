```python
#9. comparison operators - phyton
a=10
b=5

print(a == b)  # equal to
print(a != b)  #not equal to
print(a > b)   #greater than
print(a < b)   #less than
print(a >= b) #greater than or equal to
print(a <= b) #less than or equal to

```

```python
#10. Assignment operators
x=10
print(x)    #Assign 10 to x

x += 5
print(x)    #Add 5 to x: x = x + 5

x -= 3
print(x)    #Subtract 3 from x: x = x - 3

x *= 2
print(x)    #Multiply x by 2: x = x * 2

x //= 2
print(x)    #Floor divide x by 2: x = x // 2

x %= 3
print(x)    #Remainder: x = x % 3

x **= 2
print(x)    #Power: x = x ** 2

```
